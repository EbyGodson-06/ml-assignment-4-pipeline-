```python
## Qn2. Raw Data Quality Report

The raw dataset is checked for:

1. Missing values
2. Duplicate rows
3. Impossible or negative values
4. Inconsistent units
5. Inconsistent categorical labels
```

```python
# 1. Missing values
missing_values = df.isna().sum()

print("MISSING VALUES")
print("=" * 50)
print(missing_values[missing_values > 0])


# 2. Duplicate rows
duplicate_count = df.duplicated().sum()

print("\nDUPLICATE ROWS")
print("=" * 50)
print(duplicate_count)


# 3. Negative values in numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

negative_report = {}

for column in numeric_columns:
    count = (df[column] < 0).sum()

    if count > 0:
        negative_report[column] = count

print("\nNEGATIVE / IMPOSSIBLE VALUES")
print("=" * 50)

if negative_report:
    print(negative_report)
else:
    print("No negative values found")


# 4. Zero guest bookings
zero_guest_count = (
    df["adults"].fillna(0)
    + df["children"].fillna(0)
    + df["babies"].fillna(0)
).eq(0).sum()

print("\nZERO-GUEST BOOKINGS")
print("=" * 50)
print(zero_guest_count)


# 5. Zero-night bookings
zero_night_count = (
    df["stays_in_weekend_nights"]
    + df["stays_in_week_nights"]
).eq(0).sum()

print("\nZERO-NIGHT BOOKINGS")
print("=" * 50)
print(zero_night_count)


# 6. Negative ADR
negative_adr = (df["adr"] < 0).sum()

print("\nNEGATIVE ADR VALUES")
print("=" * 50)
print(negative_adr)


# 7. Inconsistent whitespace in categorical columns
categorical_columns = df.select_dtypes(include="object").columns

whitespace_report = {}

for column in categorical_columns:

    values = df[column].dropna().astype(str)

    count = (values != values.str.strip()).sum()

    if count > 0:
        whitespace_report[column] = count

print("\nCATEGORICAL WHITESPACE INCONSISTENCIES")
print("=" * 50)

if whitespace_report:
    print(whitespace_report)
else:
    print("None found")
```

```python
## Qn3. Feature Engineering

Four new features are designed from existing columns.

1. total_nights — combines weekend and weekday nights.
2. total_guests — combines adults, children and babies.
3. total_previous_bookings — combines previous canceled and non-canceled bookings.
4. adr_per_guest — calculates average daily rate per guest.

These features provide more useful information than using some
individual columns separately.
```

```python
# Create a copy
feature_df = df.copy()

# Feature 1: Total number of nights
feature_df["total_nights"] = (
    feature_df["stays_in_weekend_nights"]
    + feature_df["stays_in_week_nights"]
)

# Feature 2: Total number of guests
feature_df["total_guests"] = (
    feature_df["adults"]
    + feature_df["children"].fillna(0)
    + feature_df["babies"]
)

# Feature 3: Total previous bookings
feature_df["total_previous_bookings"] = (
    feature_df["previous_cancellations"]
    + feature_df["previous_bookings_not_canceled"]
)

# Feature 4: ADR per guest
feature_df["adr_per_guest"] = (
    feature_df["adr"] /
    feature_df["total_guests"].replace(0, np.nan)
)

# Replace infinity values with NaN
feature_df["adr_per_guest"] = (
    feature_df["adr_per_guest"].replace([np.inf, -np.inf], np.nan)
)

print("New Features Created:")
print([
    "total_nights",
    "total_guests",
    "total_previous_bookings",
    "adr_per_guest"
])

print("\nNew Shape:", feature_df.shape)
```

```python
### Qn4.2 — clean(df)

This function performs the main data cleaning operations:

- Removes duplicate rows
- Removes the highly incomplete company column
- Handles missing children values
- Handles missing country values
- Handles missing agent values
- Removes negative numeric values
- Standardizes categorical whitespace
```

```python
def clean(df):

    df = df.copy()

    # Remove duplicate rows
    df = df.drop_duplicates().reset_index(drop=True)

    # Remove company because it contains a very large proportion of missing values
    if "company" in df.columns:
        df = df.drop(columns=["company"])

    # Remove leading/trailing whitespace from categorical columns
    categorical_columns = df.select_dtypes(include="object").columns

    for column in categorical_columns:
        df[column] = df[column].astype(str).str.strip()

    # Restore missing values that became the string "nan"
    for column in categorical_columns:
        df[column] = df[column].replace("nan", np.nan)

    # Fill children missing values with 0
    if "children" in df.columns:
        df["children"] = df["children"].fillna(0)

    # Fill country missing values with Unknown
    if "country" in df.columns:
        df["country"] = df["country"].fillna("Unknown")

    # Fill agent missing values with median
    if "agent" in df.columns:
        df["agent"] = df["agent"].fillna(df["agent"].median())

    # Correct negative numeric values by replacing them with NaN
    numeric_columns = df.select_dtypes(include=np.number).columns

    for column in numeric_columns:
        df.loc[df[column] < 0, column] = np.nan

    # Fill remaining numerical missing values using median
    for column in numeric_columns:
        if df[column].isna().sum() > 0:
            df[column] = df[column].fillna(df[column].median())

    # Fill any remaining categorical missing values
    categorical_columns = df.select_dtypes(include="object").columns

    for column in categorical_columns:
        if df[column].isna().sum() > 0:
            df[column] = df[column].fillna("Unknown")

    return df
```

```python
### Qn4.3 — add_features(df)

This function creates four additional features from the cleaned dataset.
```

```python
def add_features(df):

    df = df.copy()

    # Feature 1
    df["total_nights"] = (
        df["stays_in_weekend_nights"]
        + df["stays_in_week_nights"]
    )

    # Feature 2
    df["total_guests"] = (
        df["adults"]
        + df["children"]
        + df["babies"]
    )

    # Feature 3
    df["total_previous_bookings"] = (
        df["previous_cancellations"]
        + df["previous_bookings_not_canceled"]
    )

    # Feature 4
    df["adr_per_guest"] = (
        df["adr"] /
        df["total_guests"].replace(0, np.nan)
    )

    # Replace infinite values
    df["adr_per_guest"] = (
        df["adr_per_guest"]
        .replace([np.inf, -np.inf], np.nan)
    )

    # Fill missing engineered feature values
    df["adr_per_guest"] = df["adr_per_guest"].fillna(
        df["adr_per_guest"].median()
    )

    return df
```

```python
### Qn4.4 — build_pipeline()

This function builds the preprocessing pipeline.

Numerical columns are imputed and scaled.

Categorical columns are imputed and one-hot encoded.

The target column is not included in the preprocessing pipeline.
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler


def build_pipeline():

    numeric_features = [
        "lead_time",
        "arrival_date_year",
        "arrival_date_week_number",
        "arrival_date_day_of_month",
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "adults",
        "children",
        "babies",
        "is_repeated_guest",
        "previous_cancellations",
        "previous_bookings_not_canceled",
        "booking_changes",
        "agent",
        "days_in_waiting_list",
        "adr",
        "required_car_parking_spaces",
        "total_of_special_requests",
        "total_nights",
        "total_guests",
        "total_previous_bookings",
        "adr_per_guest"
    ]

    categorical_features = [
        "hotel",
        "arrival_date_month",
        "meal",
        "country",
        "market_segment",
        "distribution_channel",
        "reserved_room_type",
        "assigned_room_type",
        "deposit_type",
        "customer_type"
    ]

    numeric_pipeline = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]
    )

    categorical_pipeline = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            (
                "onehot",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )
        ]
    )

    preprocessing = ColumnTransformer(
        transformers=[
            ("numeric", numeric_pipeline, numeric_features),
            ("categorical", categorical_pipeline, categorical_features)
        ],
        remainder="drop"
    )

    return preprocessing
```

```python
## Qn5. Full Raw-to-Model-Ready Pipeline

The complete workflow is executed in one cell:

Raw CSV
→ load_data()
→ clean()
→ add_features()
→ train/test split
→ build_pipeline()
→ X_train, X_test, y_train, y_test
```

```python
from sklearn.model_selection import train_test_split
import pandas as pd

# 1. Load raw data
raw_df = pd.read_csv("hotel_bookings.csv")

# 2. Clean data
cleaned_df = clean(raw_df)

# 3. Add engineered features
final_df = add_features(cleaned_df)

# 4. Separate target
y = final_df["is_canceled"]

# Remove target and leakage columns
X = final_df.drop(
    columns=[
        "is_canceled",
        "reservation_status",
        "reservation_status_date"
    ],
    errors="ignore"
)

# 5. Train-test split
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 6. Build preprocessing pipeline
pipeline = build_pipeline()

# 7. Fit only on training data
X_train = pipeline.fit_transform(X_train_raw)

# 8. Transform test data
X_test = pipeline.transform(X_test_raw)

# Print shapes
print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```

```python
## Qn6. Assertions for Clean Output

The following assertions verify that:

1. No NaN values remain.
2. Training and testing have the same number of columns.
3. No feature is constant.
```

```python
# Check for NaN values
assert not np.isnan(X_train).any(), "NaN values found in X_train"
assert not np.isnan(X_test).any(), "NaN values found in X_test"

# Train and test must have the same number of columns
assert X_train.shape[1] == X_test.shape[1], \
    "Train and test have different number of columns"

# Check that no feature is constant
train_feature_ranges = np.ptp(X_train, axis=0)

assert np.all(train_feature_ranges > 0), \
    "A constant feature was found"

print("All assertions passed successfully!")
print("No NaN values remain.")
print("Train and test have the same number of columns.")
print("No constant feature exists.")
```
